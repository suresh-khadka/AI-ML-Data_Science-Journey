## Logistic Regression
for categorical variable
for Clalssification variable

```python
import pandas as pd 
import numpy as  np
```

```python
x=[1,2,3,4,5]
y=[2,4,5,4,5]
import pandas as pd
df=pd.DataFrame({'X':x,
                 'Y':y})
```

```python
X=df['X']
Y=df["Y"]
X
```

```python
x_mean=np.mean(X)
```

```python
y_mean=np.mean(Y)
```

```python
# calculate m and c for eqn
deno=0
num=0
for i in range(len(X)):
    num+=(X[i]-x_mean)*(Y[i]-y_mean)
    deno+=(X[i]-x_mean) ** 2

m=num/deno
m
```

```python
c=y_mean-x_mean*m
c
```

```python
x=np.linspace(np.max(X)+2,np.min(X)-2)
y=m*x+c
```

```python
import  matplotlib.pyplot as plt
plt.plot(x,y,label="Regression line")
plt.scatter(X,Y , label="data")
plt.grid(True)
plt.legend()
```

```python
# for R2 score
ss_res=0
ss_tot=0
for i in range(len(X)):
    y_pred=m*X[i]+c
    ss_res+=(Y[i]-y_pred)**2
    ss_tot+=(Y[i]-y_mean)**2
```

```python
r2=1-(ss_res/ss_tot)
```

```python
r2
```

```python
rmse=0
for i in range(len(X)):
    y_pred=m*X[i]+c
    rmse+=(Y[i]-y_pred)**2

rmse=np.sqrt(rmse/len(Y))
```

```python
rmse
```
