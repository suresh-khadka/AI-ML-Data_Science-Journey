```python
import pandas as pd 
import seaborn as sns 
import numpy as np
import matplotlib.pyplot as plt 
%matplotlib inline
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'] 
sales = [150, 200, 300, 250, 400, 450]
```

## line plot

```python
# lilne plot
plt.plot(months,sales,color="red",marker="o", linestyle=":") 
plt.title("Monthly Sales Trends")
plt.xlabel("months")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```

## bar graph

```python

plt.bar(months,sales)
plt.title("sales Vs Months")
plt.xlabel("sales")
plt.ylabel("months")
plt.show()
```

## Hostogram

```python
plt.hist(sales,bins=5)
```

## Scatter Plot

```python

height = [150, 160, 170, 180, 190] 
weight = [50, 60, 65, 80, 90] 
plt.scatter(height,weight)
plt.title("height vs Weight")
```

```python
plt.pie(sales, autopct="%1.2f%%")
```
