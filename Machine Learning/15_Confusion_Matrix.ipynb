```python
import pandas as pd
```

```python
df = pd.read_csv('heart_disease_data.csv')
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
lr_model=LogisticRegression()
Dtree_model=DecisionTreeClassifier()
```

```python
lr_model.fit(X_train , y_train)
Dtree_model.fit(X_train , y_train)
y_pred_lr=lr_model.predict(X_test)
y_pred_tree=Dtree_model.predict(X_test)
```

### Accracy Score 

```python
from sklearn.metrics import accuracy_score
print("Accuracy of LR model " ,accuracy_score(y_test,y_pred_lr) )
print("-"*80)
print("Accuracy of Tree model " ,accuracy_score(y_test,y_pred_tree) )
```

### Confusion Matrix

```python
print("-"*80)
from sklearn.metrics import confusion_matrix
print(" confsion Matrix " )
print(confusion_matrix(y_test,y_pred_lr))
print("-"*80)
print(" confsion Matrix " )
print(confusion_matrix(y_test,y_pred_tree))
print("-"*80)
```

### Precision

```python
from sklearn.metrics import precision_score
print("Precision Score for Logistic Regression " )
print(precision_score(y_test,y_pred_lr))
print("-"*80)
print("precision Score for Decision Tree" )
print(precision_score(y_test,y_pred_tree))
```

### Recall Score

```python
from sklearn.metrics import recall_score
print("Recall Score for Logistic Regression " )
print(recall_score(y_test,y_pred_lr))
print("-"*80)
print("recall_score for Decision Tree" )
print(recall_score(y_test,y_pred_tree))
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy of LR model " ,accuracy_score(y_test,y_pred_lr) )
print("-"*80)
print("Accuracy of Tree model " ,accuracy_score(y_test,y_pred_tree) ) 
```

### F1 Score

```python
from sklearn.metrics import f1_score
print("F1 Score for Logistic Regression " )
print(f1_score(y_test,y_pred_lr))
print("-"*80)
print("F1 Score for Decision Tree" )
print(f1_score(y_test,y_pred_tree))
```

```python
df=pd.read_csv("mnist1.5k.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0:1],test_size=0.2,random_state=2)
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
lr_model=LogisticRegression()
Dtree_model=DecisionTreeClassifier() 
lr_model.fit(X_train , y_train)
Dtree_model.fit(X_train , y_train)
y_pred_lr=lr_model.predict(X_test)
y_pred_tree=Dtree_model.predict(X_test)
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred_lr),columns=list(range(0,10)))
```

```python
from sklearn.metrics import precision_score 
from sklearn.metrics import precision_score
print("Precision Score for Logistic Regression " )
print(precision_score(y_test,y_pred_lr , average=None))
print(precision_score(y_test,y_pred_lr , average="macro"))
print("-"*80)
print("precision Score for Decision Tree" )
print(precision_score(y_test,y_pred_tree , average="weighted"))
```

```python
print("recall_score for Decision Tree" )
print(recall_score(y_test,y_pred_tree , average="weighted"))
print("F1 Score for Decision Tree" )
print(f1_score(y_test,y_pred_tree , average="macro"))
```

### Classification Report

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred_lr))
```
