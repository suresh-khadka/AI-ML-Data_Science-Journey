```python
import pandas as pd 
```

```python
df=pd.read_csv("mnist1.5k.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0:1],test_size=0.2,random_state=2)
from sklearn.linear_model import LogisticRegression

lr_model=LogisticRegression()

lr_model.fit(X_train , y_train)

y_pred_lr=lr_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred_lr))
```

```python
from sklearn.metrics import precision_score
print("Precision Score for Logistic Regression " )
print('-'*100)
print("Average None precision score", end="   ")
print(precision_score(y_test,y_pred_lr , average=None))
print('-'*100)
print("weighted precision score", end="   ")
print(precision_score(y_test,y_pred_lr , average="macro"))
```

```python
from sklearn.metrics import precision_score
from sklearn.metrics import classification_report, f1_score
print('-'*100)
print("Classification Report : ")
print(classification_report(y_test,y_pred_lr))
print('-'*100) 
print('-'*100)

from sklearn.metrics import accuracy_score
print("Accracy Score for Logistic Regression " )
print("Accuracy score", end="   ")
print(accuracy_score(y_test,y_pred_lr ))
print('-'*100)
print('-'*100)

print("Precision Score for Logistic Regression " )
print("Average None precision score", end="   ")
print(precision_score(y_test,y_pred_lr , average=None))
print("weighted precision score", end="   ")
print(precision_score(y_test,y_pred_lr , average="macro"))
print('-'*100) 
print('-'*100)
print("F1 Score for Logistic Regression " )
print("Average None F1 score", end="   ")
print(f1_score(y_test,y_pred_lr , average=None))
print("weighted F1 score", end="   ")
print(f1_score(y_test,y_pred_lr , average="macro"))
print("-"*80)
```
