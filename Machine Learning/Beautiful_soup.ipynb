### for real Webbsite

```python
url='https://www.worldometers.info/coronavirus/'
response=requests.get(url)
soup=BeautifulSoup(response.text,'html.parser')
soup.prettify()
```

```python
tag=soup.find('table',id='main_table_countries_today')
```

```python
print(tag)
```

```python
tag_new=tag.find_all('tr')
tag_new
```

```python
lst=[]
for i in tag_new:
    lst.append(i.text.strip())
```

```python
lst
```

```python
data=[]
for j in lst:
    data.append(j.split('\n'))

len(data)
```

```python
data
```

```python
data = [[x for x in row if x != ''] for row in data]
```

```python
cols=data[0]
```

```python
cols.remove('#')
cols
```

```python
data=data[1:]
data
```

```python
import  pandas as pd 
df=pd.DataFrame(data,columns=cols)
```
