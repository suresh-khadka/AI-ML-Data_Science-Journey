```python
import matplotlib.pyplot as plt
from sklearn.inspection import DecisionBoundaryDisplay

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

# 1. Plot Linear Kernel Boundary
DecisionBoundaryDisplay.from_estimator(
    svc_linear, 
    X_train, 
    response_method="predict", 
    alpha=0.3, 
    cmap=plt.cm.coolwarm, 
    ax=ax1
)
ax1.scatter(X_train[:, 0], X_train[:, 1], c=y_train, edgecolor="k", cmap=plt.cm.coolwarm)
ax1.set_title("SVC with Linear Kernel\n(Cannot separate nested shapes)")
ax1.set_xlabel("Feature 1")
ax1.set_ylabel("Feature 2")

# 2. Plot RBF Kernel Boundary
DecisionBoundaryDisplay.from_estimator(
    svc_rbf, 
    X_train, 
    response_method="predict", 
    alpha=0.3, 
    cmap=plt.cm.coolwarm, 
    ax=ax2
)
ax2.scatter(X_train[:, 0], X_train[:, 1], c=y_train, edgecolor="k", cmap=plt.cm.coolwarm)
ax2.set_title("SVC with RBF Kernel\n(Easily separates nested shapes)")
ax2.set_xlabel("Feature 1")
ax2.set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```
