```python
import pandas as pd 
import numpy as np 
```

```python
df=pd.read_csv("titanic_01-handling-missing-data.csv")
df.head()
```

```python
df['number'].unique()
```

```python
df.isnull().sum()
```

```python
df['number_integer']=pd.to_numeric(df['number'],downcast='integer' , errors='coerce')
df['number_categorical']=np.where(df['number_integer'].isnull() , df['number'] , np.nan)
```

Smart Feature Engineering for Cabin

Extract first letter:

df['Cabin_Type'] = df['Cabin'].str[0]

```python
df['cabin_num']=df["Cabin"].str.extract('([\d]+)')
df['cabin_cat']=df['Cabin'].str[0]
df.head(10)
```

Smart Feature Engineering for Ticket

Ticket values:

A/5 21171
PC 17599
STON/O2. 3101282

You can extract:

Ticket Prefix
df['Ticket_Prefix'] = df['Ticket'].str.extract('([A-Za-z./]+)')

```python
# extract ticket prefix
df['ticket_prefix'] = df['Ticket'].str.extract('([A-Za-z0-9./]+)')

# fill missing prefixes
df['ticket_prefix'] = df['ticket_prefix'].fillna('NoPrefix')

# extract last numeric part
df['ticket_number'] = df['Ticket'].str.extract('(\d+$)')

# convert to numeric
df['ticket_number'] = pd.to_numeric(df['ticket_number'])
```
