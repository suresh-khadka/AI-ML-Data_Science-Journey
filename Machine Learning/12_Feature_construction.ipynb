```python
import pandas as pd 
import numpy as np 
```

```python
df=pd.read_csv('titanic.csv')
df.head()
```

```python
df['Family_size']=df['Parch'] + df['SibSp']+ 1 
df.head()
```

```python
#introduce family type
# 0-small family 
# 1 medium family 
# 2 large family

def Family_type(data):
    if data == 1:
        return 0
    elif data <4:
        return 1
    else:
        return 2
```

```python
df["Family_type"]=df['Family_size'].apply(Family_type)
```

### 1. Improves Model Performance
Good features help models discover patterns more easily.
Example:
Birth Year → Age
Age is usually more useful than birth year.

2. Helps Simple Models Perform Better
Sometimes:
Good Features + Logistic Regression
outperform
Poor Features + XGBoost
Feature quality often matters more than algorithm complexity.

3. Reveals Hidden Relationships
Example:
FamilySize = SibSp + Parch + 1
The model may learn family behavior better than from separate columns.

4. Makes Data More Meaningful
Example:
BMI = Weight / Height²
BMI carries more health information than weight or height alone.

5. Can Reduce Noise
Instead of exact age:
18, 19, 20, 21...
Create:
Child, Adult, Senior
which may generalize better.

6. Helps Handle Nonlinear Relationships
Example:x²
allows linear models to learn curved relationships.

7. Incorporates Domain Knowledge
Example:
Finance:
Debt-to-Income Ratio
These often become highly predictive features.

8. Better Interpretability
Features like:
Average Purchase Amount
Customer Lifetime Value
are easier to explain than raw transaction data.

Disadvantages of Feature Creation
1. Can Cause Overfitting
Creating too many features may cause the model to memorize training data.
Example:
x
x²
x³
x⁴
...
x²⁰
Model becomes overly complex.

2. Increases Dimensionality
More features mean:
More memory
More storage
Longer training time
This is called the curse of dimensionality.

3. Requires Domain Knowledge
Without understanding the problem:
Useful Feature Creation = Difficult
Bad features can hurt performance.

4. Risk of Data Leakage
A common beginner mistake.
Example:
Using future information to create a feature.
Exam Score → Predict Exam Pass
This leaks the answer into the model.

5. More Complex Pipeline
Instead of:
Raw Data → Model
you now have:
Raw Data
   ↓
Feature Creation
   ↓
Encoding
   ↓
Scaling
   ↓
Model
Maintenance becomes harder.

6. Some Features May Be Redundant
Example:
Area = Length × Width
Perimeter = 2(Length + Width)
Sometimes new features add little information.

7. Increased Computation Cost
Complex features may require:
groupby operations
aggregations
text processing
date calculations
which can slow training.
